```python
def count_positive_negative(numbers):
    """
    Return a dictionary with positive, negative and zero counts.
    """
    result = {'positive' : 0, 'negative' : 0, 'zero' : 0}
    for number in numbers:
        if number > 0 :
            result['positive']+=1
        elif number < 0 :
            result['negative']+=1
        else :
            result['zero']+=1
    return result

numbers = [4, -2, 0, 7, -8, 0]
result=count_positive_negative(numbers)
print(result)
```

```python
# Input

# [1,2,3,4,5,6,7]
# batch_size = 3

# Output

# [
# [1,2,3],
# [4,5,6],
# [7]
# ]
def batch(items: list[int], batch_size: int) -> list[list[int]]:
    ''' 
    Function to send batches of data
    Arguments: List of items
    Returns: List of Integer Lists
    '''
    result = []
    if batch_size == 0 or len(items) == 0 or batch_size > len(items) :
        return 'Invalid Input'
    for i in range(0,len(items),batch_size) :
        result.append(items[i:i+batch_size])
    return result
print(batch([1,2,3,4,5,6,7], 3))
```

```python
def calculate_total(numbers: list[int]) -> int:
    total = 0
    for i in numbers:
        total += i
    return total
print(calculate_total([10, 20, 30, 40]))

def count_greater_than_10(numbers: list[int]) -> int:
    count = 0
    for i in numbers:
        if i > 10 : count += 1
    return count
print(count_greater_than_10([5, 15, 20, 3, 12]))

import sys
def find_largest(numbers: list[int]) -> int | None:
    if not numbers:
        return None
    largest = -sys.largestsize - 1
    for i in numbers:
        if i > largest:
            largest = i
    return largest
print(find_largest([10, 10, 10]))

def find_first_even(numbers: list[int]) -> int | None:
    for i in numbers:
        if i % 2 == 0:
            return i
    return None
print(find_first_even([1, 3, 5]))
    
```

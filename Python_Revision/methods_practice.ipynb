```python
'''List'''
numbers = [10, 20, 30, 20, 40]
numbers.append(50)
numbers.extend([60,70])
numbers.insert(1,15)
numbers.remove(20)
numbers.pop()
print(numbers.count(20))
print(numbers.index(30))
numbers.sort(reverse=True)
print(numbers)

numbers = [1, 2, 3, 4, 5, 6, 7, 8]
print([i for i in numbers if i%2 == 0])
```

```python
text = 'banana'
counts = {}

for char in text:
    counts[char] = counts.get(char,0) + 1

print(counts)

person = {
    "name": "Vilva",
     "age": 22,
    "city": "Kavali"
}
print(person.get('name'))
print(person.get('salary',0))
person['age'] = 23
print(person)
person['role'] = 'Python Developer'
print(person)

print(person.keys())
print(person.values())

for i in person.keys():
    print(i)

for i in person.values():
    print(i)

for key,value in person.items():
    print(key,' : ',value)

person.pop('city')
print(person)
```

```python
#Sets
#  add()       → one item
# update()    → multiple
# remove()    → remove, error if missing
# discard()   → remove, no error
# in          → membership
# |           → union
# &           → intersection
# -           → difference
# ^           → symmetric difference
```

```python
# Input
 
# apple apple apple banana banana cat
 
# Output
 
# apple3 banana2 cat1

def compress_words(text: str) -> str:
    counts={}
    result=''
    for word in text.split():
        counts[word] = counts.get(word,0)+1

    for key,value in counts.items():
        result += str(key) + str(value)+' '
    
    return result
print(compress_words('apple apple apple banana banana cat'))
```

```python
# Move all zeros to the end of a list while maintaining the order of non-zero elements.
# Input:  [0, 1, 0, 3, 12]
# Output: [1, 3, 12, 0, 0]
def zero_swap(numbers: list[int]) -> list[int]:
    result = []
    for i in numbers:
        if i != 0:
            result.append(i)
    result.extend([0 for i in range(len(numbers)-len(result))])
    return result
print(zero_swap([0, 1, 0, 3, 12]))
```

```python
# Given a list of strings, group all anagrams together and return the result as a list of lists.
 
# Two strings are considered anagrams if they contain the same characters with the same frequencies, regardless of their order.
# input:
# words = ["eat", "tea", "tan", "ate", "nat", "bat"]
 
# output:
# [
#     ["eat", "tea", "ate"],
#     ["tan", "nat"],
#     ["bat"]
# ]
def anagrams(words: list[str]) -> list[list[str]]:
    sorted_dict = {}
    for word in words:
        key = ''.join(sorted([i for i in word]))
        if sorted_dict.get(key) is None:
            sorted_dict[key] = []
        sorted_dict[key].append(word) 
    return list(sorted_dict.values())

result = anagrams(["eat", "tea", "tan", "ate", "nat", "bat"])
print(result)
```

```python
# Merge two dictionaries, summing shared keys

# d1 = {"a": 10, "b": 20, "c": 30}
# d2 = {"b": 5, "c": 10, "d": 15}

# Expected: {"a": 10, "b": 25, "c": 40, "d": 15}
def dict_merge(dict1: dict[str,int], dict2: dict[str,int]) -> dict[str,int]:
    result = dict1
    for key,value in dict2.items():
        if key not in result:
            result[key] = value
        else:
            result[key] += value
    return result

print(dict_merge({"a": 10, "b": 20, "c": 30},{"b": 5, "c": 10, "d": 15})) 
```

```python
# Flatten a nested list

# nested = [1, [2, 3, [4, 5]], 6, [7]]

# Expected: [1, 2, 3, 4, 5, 6, 7]

def flatten(nested_list: list[list]) -> list[any]:
    result = []
    for i in nested_list:
        if isinstance(i,list):
            result.extend(flatten(i))
        else:
            result.append(i)
    return result

print(flatten([1, [2, 3, [4, 5]], 6, [7]]))
```

```python
# Find the second largest number
# Input:  [10, 5, 8, 10, 3]
# Output: 8
def second_largest(numbers: list[int]):
    largest=0
    second_largest=0
    for i in numbers:
        if i == largest:
            continue
        elif i > largest:
            second_largest = largest
            largest = i
        elif i > second_largest:
            second_largest = i
    return second_largest
print(second_largest([10, 5, 8, 10, 3]))
```

```python
# Two numbers with maximum product
# a = [1, 5, 3, 6, 2]
 
# [-10, -5, 1, 2]
def max_product(numbers: list[int]) -> list[int,int]:
    max_product = 0
    result = []
    for i in range(len(numbers)):
        for j in range(i+1, len(numbers)):
            product = numbers[i]*numbers[j]
            if product > max_product:
                max_product = product
                result = [numbers[i], numbers[j]]
    return result
print(max_product([1, 5, 3, 6, 2]))

def optimized_max_product(numbers: list[int]) -> list[int,int]:
    numbers.sort()

    left_product = numbers[0] * numbers[1]
    right_product = numbers[-1] * numbers[-2]

    if left_product > right_product:
        return [numbers[0],numbers[1]]
    return [numbers[-1],numbers[-2]]

print(optimized_max_product([-10, -5, 1, 2]))
```

```python
# a = [-1, 0, 1, 2, -1, -4]
# Find triplets whose sum is 0.
# Output:
# [-1, -1, 2][-1, 0, 1]
def triplet_finder(numbers: list[int],target: int) -> list[list[int]]:
    result = []
    numbers.sort()

    for i in range(len(numbers)-2):

        if i > 0 and numbers[i] == numbers[i-1]:
            continue

        l = i+1
        r = len(numbers)-1
        while l < r:
            total = numbers[i]+numbers[l]+numbers[r]
            if total == target:
                result.append([numbers[i],numbers[l],numbers[r]])
                l+=1
                r-=1
            elif total < target:
                l+=1
            else:
                r-=1
    return result

print(triplet_finder([-1, 0, 1, 2, -1, -4],0))
```
